1. | SQL | NoSQL |
|---|---|
| Uses **tables, rows, and columns** | Uses **documents, key-value, graphs, etc.** |
| Has a **fixed schema** | Has a **flexible schema** |
| Uses **SQL** for queries | Query language varies by database |

| Traditionally scales **vertically** | Traditionally scales **horizontally** |
| Best for **structured data** | Good for **semi/unstructured data** |

2. ### Why MongoDB is a Good Choice for Modern Applications

MongoDB is a **NoSQL, document-based database** that is well suited for modern applications because:

- **Flexible schema:** Data structure can change without redesigning tables.
- **Scalable:** Supports horizontal scaling across multiple servers.
- **High performance:** Handles large volumes of read/write operations efficiently.
- **JSON-like documents:** Stores data in BSON format, which is easy to work with in modern programming languages.
- **Easy development:** Developers can store application objects directly as documents.
- **High availability:** Supports replication and automatic failover.

**In short:** MongoDB is popular because it provides **flexibility, scalability, high performance, and developer-friendly data storage**.

3.### Collections in MongoDB

A **collection** in MongoDB is a group of related **documents**, similar to a **table in SQL**.

- Collections store documents in **BSON (Binary JSON)** format.
- Unlike SQL tables, collections **do not require a fixed schema**.
- Documents within the same collection can have different fields.
- A collection is created automatically when data is first inserted.

**Example:**

```text
Database: Company
Collection: Employees

{
  "name": "Rahul",
  "age": 25,
  "department": "IT"
}
```

**In short:**  
**Database → Collection → Documents → Fields**

Collection in MongoDB ≈ Table in SQL.

4. ### MongoDB High Availability Through Replication

MongoDB uses **replica sets** to provide high availability.

- A **primary node** handles read and write operations.
- **Secondary nodes** maintain copies of the primary's data.
- If the primary fails, a **secondary is automatically elected as the new primary**.
- This ensures **continuous availability and data redundancy**.

**In short:** MongoDB uses replication and automatic failover to keep the database available even when a server fails.

6. ### Indexes in MongoDB

An **index** is a data structure that helps MongoDB find documents faster without scanning the entire collection.

**Benefits:**
- Speeds up **query and search operations**.
- Improves performance for **sorting and filtering**.
- Reduces the number of documents MongoDB needs to examine.

**Example:** An index on `email` allows MongoDB to quickly find a customer by email instead of checking every document.

**In short:** Indexes improve **query performance and efficiency**, especially for large collections.

7. ### Stages of MongoDB Aggregation Pipeline

The **aggregation pipeline** processes documents through a series of stages to filter, transform, and analyze data.

Common stages include:

1. **$match** – Filters documents based on conditions.
2. **$group** – Groups documents and performs calculations like sum, average, and count.
3. **$sort** – Sorts the results.
4. **$project** – Selects or modifies fields.
5. **$limit** – Limits the number of documents returned.
6. **$skip** – Skips a specified number of documents.
7. **$unwind** – Deconstructs an array into separate documents.

**In short:** The pipeline moves data through multiple stages to **filter, transform, group, and analyze documents efficiently**.

8. ### Sharding vs Replication in MongoDB

**Sharding** is the process of **splitting large amounts of data across multiple servers (shards)** to improve scalability and distribute workload.

**Replication** creates **multiple copies of the same data** on different servers to provide high availability and fault tolerance.

| Sharding | Replication |
|---|---|
| Distributes data across servers | Copies the same data across servers |
| Improves **scalability and performance** | Improves **availability and reliability** |
| Used for very large datasets | Used for backup/failover |
| Data is divided among shards | Each replica maintains a copy |

**In short:**  
**Sharding = divide data for scalability.**  
**Replication = copy data for availability.**

10. ### ACID Properties in MongoDB Transactions

ACID ensures that database transactions are **reliable and consistent**.

1. **Atomicity** – A transaction is completed fully or not at all.
2. **Consistency** – Data remains valid and follows defined rules before and after a transaction.
3. **Isolation** – Multiple transactions do not interfere with each other.
4. **Durability** – Once a transaction is committed, its changes are permanently stored.

**In short:** ACID ensures MongoDB transactions are **reliable, consistent, and safe**.

11. ### MongoDB `explain()` Function

The **`explain()`** function is used to **analyze how MongoDB executes a query**.

It helps you:
- Check whether **indexes** are being used.
- Understand the **query execution plan**.
- Identify slow or inefficient queries.
- Improve **query performance**.

**In short:** `explain()` helps developers **find and optimize inefficient MongoDB queries**.

12. ### MongoDB Schema Validation

MongoDB uses **schema validation rules** to control the structure and data types of documents stored in a collection.

- Uses **JSON Schema** to define validation rules.
- Can specify **required fields, data types, ranges, and formats**.
- Invalid documents can be **rejected** when inserted or updated.
- It provides flexibility while maintaining **data quality and consistency**.

**In short:** MongoDB allows a flexible schema but can use validation rules to ensure **correct and consistent data**.

13. ### Primary vs Secondary Node in MongoDB

- **Primary Node:** Handles all **write operations** and, by default, read operations. It is the main node in a replica set.
- **Secondary Node:** Maintains a **copy of the primary's data** by replicating its operations. It can serve read operations if configured to do so.

**In short:**  
**Primary = main read/write node**  
**Secondary = replicated backup node used for redundancy and, optionally, reads**.

14. ### MongoDB Security Mechanisms

MongoDB provides several security features to protect data:

- **Authentication** – Verifies user identity.
- **Authorization (RBAC)** – Controls what users can access or modify.
- **Encryption** – Protects data both **in transit and at rest**.
- **TLS/SSL** – Secures network communication.
- **Auditing** – Tracks database activities and user actions.
- **Network controls** – Restricts access to trusted networks and IP addresses.

**In short:** MongoDB protects data through **authentication, access control, encryption, auditing, and network security**.

15. ### Embedded Documents in MongoDB

**Embedded documents** are documents stored **inside another document** as nested objects or arrays.

**Example:**
```json
{
  "name": "Rahul",
  "address": {
    "city": "Delhi",
    "pincode": 110001
  }
}
```

Here, `address` is an embedded document within the customer document.

**When to use:**
- When related data is **frequently accessed together**.
- When the embedded data has a **one-to-one or one-to-few relationship**.
- When you want to reduce the need for **joins or multiple queries**.

**In short:** Embedding keeps related data together, making data retrieval **faster and simpler**.

16. ### MongoDB `$lookup` Stage

The **`$lookup`** stage is used to **combine documents from two collections**, similar to a **JOIN in SQL**.

- Matches related fields between collections.
- Adds matching documents to the aggregation result.
- Useful for analyzing related data stored in different collections.

**In short:** `$lookup` performs a **JOIN-like operation between MongoDB collections**.

19. ### MongoDB Transactions vs SQL Transactions

| MongoDB | SQL |
|---|---|
| Supports **multi-document transactions** | Supports transactions across multiple rows/tables |
| Uses **documents and collections** | Uses **tables and relationships** |
| Transactions are typically used when multiple documents must change together | Transactions are commonly used for complex relational operations |
| Uses MongoDB-specific transaction APIs | Uses standard SQL commands like `BEGIN`, `COMMIT`, and `ROLLBACK` |
| Flexible schema | Usually fixed schema |

**In short:** Both support **ACID transactions**, but MongoDB transactions work with **documents and collections**, while SQL transactions work primarily with **rows and tables**.

20. ### Capped vs Regular Collections in MongoDB

| Capped Collection | Regular Collection |
|---|---|
| Has a **fixed size** | Size can **grow dynamically** |
| Automatically removes **oldest documents** when full | Documents remain until explicitly removed |
| Maintains **insertion order** | No guaranteed natural insertion order |
| Cannot grow beyond its set limit | Can grow as needed |
| Useful for **logs, caching, and temporary data** | Used for **general-purpose data storage** |

**In short:** Capped collections are **fixed-size and automatically overwrite old data**, while regular collections **grow dynamically and retain data until it is deleted**.

21. ### `$match` Stage in MongoDB

The **`$match`** stage is used to **filter documents** in an aggregation pipeline based on specified conditions.

**Example:**
```javascript
{ $match: { age: { $gt: 25 } } }
```

This returns only documents where `age` is greater than 25.

**In short:** `$match` **filters data** and can improve performance by reducing the number of documents processed by later stages.

22. ### Securing Access to a MongoDB Database

- **Authentication:** Require users to log in with valid credentials.
- **Authorization (RBAC):** Assign roles and permissions based on user requirements.
- **Encryption:** Encrypt data **at rest and in transit** using TLS.
- **Network Security:** Restrict access using IP allowlists, firewalls, or private networks.
- **Auditing:** Monitor and record user activities and database access.
- **Strong Passwords:** Use strong passwords and rotate credentials regularly.

**In short:** Secure MongoDB using **authentication, authorization, encryption, network controls, and auditing**.

23. ### MongoDB WiredTiger Storage Engine

**WiredTiger** is MongoDB’s default storage engine. It manages how MongoDB **stores, reads, and writes data** on disk.

**Why it is important:**
- Supports **document-level concurrency** for better performance.
- Provides **compression** to reduce storage usage.
- Supports **ACID transactions**.
- Uses **journaling** to help recover data after failures.
- Handles large workloads efficiently.

**In short:** WiredTiger is important because it provides MongoDB with **high performance, concurrency, reliability, and efficient storage**.
